## Exploring the data

### USDA data set
[Unemployment and median household income for the United States, States, and counties, 2000–23](https://www.ers.usda.gov/data-products/county-level-data-sets/county-level-data-sets-download-data)

```python
df_USDA = pd.read_csv('Data/CountyIncomeUSDA/Unemployment2023.csv')
```

```python
df_USDA.info()
```

Federal Information Processing System (FIPS) codes identify geographic areas by state and county.
[FIPS Code details](https://transition.fcc.gov/oet/info/maps/census/fips/fips.txt)

```python
df_USDA.head(10)
```

```python
# Pull attributes so I can explore 
df_attributes = df_USDA['Attribute']
df_attributes.value_counts()
```

```python
# Remove year so I can see genneral types of info available
df_attributes = df_attributes.replace(r'_[0-9]+', '', regex=True).str.strip()
df_attributes.value_counts()
```
